#### if문 - 조건문

```python
#기본 if문
a = 0
if a==0 : 
    print("조건 성공")
```

```python
#if~else문
score  = 80
if score >=80 :
    print('합격')
else :
    print("불합격")
```

```python
#elif문
score = 80
if score >=90 :
    print("A학점")
    print("합격!")
elif score >=80 :
    print("B학점")
    print("합격!")
elif score >=70 :
    print("C학점")
    print("불합격..")
elif score >=60 :
    print("D학점")
    print("불합격..")
else :
    print("F학점")
    print("불합격..")
```

```python
#중첩 if문
score = 89
if score >=90 :
    if score >= 95:
        print("A+학점")
    else :
        print("A학점")
    print("합격!")
elif score >=80 :
    print("B학점")
    print("합격!")
elif score >=70 :
    print("C학점")
    print("불합격..")
elif score >=60 :
    print("D학점")
    print("불합격..")
else :
    print("F학점")
    print("불합격..")
```

```python
temp = 24
is_raining = True

if temp >=25 :
    print("더움")
    if is_raining :
        print("비옴")
    else :
        print("비안옴")
else :
    print("안더움")
    if is_raining :
        print("비옴")
    else :
        print("비안옴")


temp = 27   
is_raining = True
if temp >= 25 and is_raining:  # is_raining == True
    print('더운데 비까지 옴')
elif temp >= 25 and not is_raining :
    print('더운데 비는 안옴')
elif temp < 25 and is_raining:
    print('안 더운데 비는 옴')
else:
    print('안 더운데 비도 안옴')
```

#### 반복문

##### while문
if 문처럼 조건이 참인 동안 계속해서 로직을 반복 실행. 조건이 거짓이 되면 반복을 종료

```python
count = 1
while count < 5 :
    print(f"반복 {count}번째중")
    count += 1
print("반복종료")
```

```python
while True :
    user_input = input('종료하려면 q를 누르세요')
    if(user_input == 'q') :
        print("프로그램 종료")
        break #반복문 탈출
    else :
        print(f'입력값 : {user_input}')
```

```python
password = ''
while password != '1q2w3e4r' :
    password = input('비밀번호를 입력해주세요')
    if(password == '1q2w3e4r') :
        print("접속완료")
    else :
        print('비밀번호가 틀렸습니다')
        continue
```

```python
# continue을 이용해 현재 반복만 건너뛰고, 바로 다음 반복으로 넘어갑니다
number = 0
while number <=5 :
    number += 1
    if number == 3 :
        continue
    print(number)
```

##### for문

```python
fruits =['사과','바나나','포도','오렌지','망고']
for fruit in fruits :
    print(f'이번 과일은 {fruit}')
```

```python
for i in range(5):  # range(5)  -> range(0:5)
    print(i)
print('---')
for i in range(1,6):
    print(i)
```

```python
for i in range(2,10):
    print(f'{i}단')
    for j in range(1,11) :
        print(f'{i} x {j} = {i*j}')
    print('----------')
```

```python
numbers = [i for i in range(1,11)]
print(numbers)

for num in numbers :
    print(f'number = {num}')
    if num == 5 :
        break
   
print('-------------')
for num in numbers :
    if num % 3 == 0 :
        continue
    print(f'number = {num}')
```
