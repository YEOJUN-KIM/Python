# 성적판별하기

- 사용자로부터 점수를 입력받음
- 점수를 정수형으로 변환
- 다음 규칙에 따라 판별 후 출력
    - 90점 이상 : A학점
    - 80점 이상 : B학점
    - 70점 이상 : C학점
    - 그외 : F학점

```python
while True:
    scr = input('점수를 입력해주세요: ')
    try:
        score = int(scr)
        if(score >100 or score <0) :
            print('올바른 점수를 입력해주세요!')
            continue
        break
    except ValueError:
        print('숫자를 입력해주세요!')

if score >=90 :
    print("A학점")
elif score >=80 :
    print("B학점")
elif score >=70 :
    print("C학점")
else :
    print("F학점")
```
